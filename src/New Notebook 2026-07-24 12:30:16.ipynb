```python
from pyspark.sql import functions as F

data = [
    (1, "Alice", 25, 50000),
    (2, "Bob", 30, 70000),
    (3, "Charlie", 22, 40000)
]

df = spark.createDataFrame(data, ["id", "name", "age", "salary"])
df.show()
```

```python
df.select("name", "salary").show()
```

```python
df.filter(F.col("age") > 25).show()
```

```python
df.withColumn("salary_after_bonus", F.col("salary") * 1.1).show()
```

```python
df.withColumn(
    "age_group",
    F.when(F.col("age") < 25, "Young")
     .when(F.col("age") <= 30, "Mid")
     .otherwise("Senior")
).show()
```

```python
employees = spark.createDataFrame(
    [(1, "Alice", 101), (2, "Bob", 102), (3, "Charlie", 101)],
    ["emp_id", "emp_name", "dept_id"]
)

departments = spark.createDataFrame(
    [(101, "HR"), (102, "IT")],
    ["dept_id", "dept_name"]
)

employees.join(departments, on="dept_id", how="inner").show()
```

```python
employees.join(departments, on="dept_id", how="left").show()
employees.join(departments, on="dept_id", how="right").show()
employees.join(departments, on="dept_id", how="full").show()
```

```python
df.groupBy("age").agg(
    F.count("*").alias("count"),
    F.avg("salary").alias("avg_salary"),
    F.max("salary").alias("max_salary")
).show()
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("salary", IntegerType(), True)
])

df_schema = spark.createDataFrame(data, schema)
df_schema.printSchema()
```
