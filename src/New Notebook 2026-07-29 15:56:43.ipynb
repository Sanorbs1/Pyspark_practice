```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("BroadcastAccumulatorExample").getOrCreate()

data = [
    ("u1", 100.0, "IT"),
    ("u2", -50.0, "HR"),      # invalid: negative amount
    ("u3", 200.0, "FIN"),
    ("u4", 300.0, "UNKNOWN"), # invalid: unknown department
]

transactions_df = spark.createDataFrame(data, ["user_id", "amount", "department_code"])
transactions_df.show()
```

```python
# Small reference data: department_code -> department_name
dept_lookup = {
    "IT": "Information Technology",
    "HR": "Human Resources",
    "FIN": "Finance",
}

# Broadcast to all executors
bc_dept_lookup = spark.sparkContext.broadcast(dept_lookup)
```
