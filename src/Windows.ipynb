```python
data = [
    ("A", "Alice", 1000),
    ("A", "Bob", 1500),
    ("A", "Charlie", 1200),
    ("B", "David", 2000),
    ("B", "Eve", 1800),
    ("B", "Frank", 2500)
]

df = spark.createDataFrame(data, ["dept", "name", "salary"])
df.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql import functions as F

w = Window.partitionBy("dept").orderBy(F.col("salary").desc())
```

```python
df.withColumn("row_num", F.row_number().over(w)).show()
```

```python
df.withColumn("rank", F.rank().over(w)).show()
```

```python
df.withColumn("dense_rank", F.dense_rank().over(w)).show()
```

```python
df.withColumn("prev_salary", F.lag("salary", 1).over(w)).show()
```

```python
df.withColumn("next_salary", F.lead("salary", 1).over(w)).show()
```
