```python
from pyspark.sql import functions as F

data = [
    (1, "  Alice ", "alice@example.com", "2026-07-01", 100.5, "HR", "New York"),
    (2, "Bob", "bob@test.com", "2026-07-02", 250.0, "IT", "Boston"),
    (3, "Charlie", "charlie@example.com", "2026-07-03", 175.75, "IT", "Boston"),
    (4, "David", "david@gmail.com", "2026-07-04", 300.0, "Finance", "Chicago"),
    (5, "Eve", "eve@yahoo.com", "2026-07-05", 125.0, "HR", "New York")
]

df = spark.createDataFrame(data, ["id", "name", "email", "date_str", "salary", "dept", "city"])
```

```python
df.show()
df.show(5, truncate=False)
```
